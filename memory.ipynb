```python
from langchain_community.document_loaders import PyPDFLoader,Docx2txtLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings
from typing import List
from langchain_core.documents import Document
```

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 1000,
    chunk_overlap = 200,
    length_function = len
)

docx_loader = Docx2txtLoader("docs/Football_Prediction_2026.docx")
documents = docx_loader.load()
print(len(documents))

splits = text_splitter.split_documents(documents)

print(f"Split the documents into {len(splits)} chunks.")
```

```python
embeddings = OpenAIEmbeddings()

document_embeddings = embeddings.embed_documents([split.page_content for split in splits])
print(f"Created embeddings for {len(document_embeddings)} document chunks.")
```

```python
from langchain_chroma import Chroma
# ১. ডেটা কোথায় সেভ হবে সেই ফোল্ডারের নাম বা পাথ দিন
persist_directory = "./chroma_db"
embedding_function = OpenAIEmbeddings()
collection_name = 'my_collection'

vectorstore = Chroma.from_documents(
    collection_name=collection_name,
    documents=splits,
    embedding=embedding_function,
    persist_directory=persist_directory
)
print(f"Vector store created and persisted to '{persist_directory}'")
```

```python
# ১. আপনার তৈরি করা vectorstore থেকে কালেকশনের তথ্য নিন
collection_data = vectorstore.get(include=['documents', 'metadatas', 'embeddings'])
collection_data
```

```python
# ২. ডাটাবেজে মোট কয়টি টুকরো বা ভেক্টর আছে তা দেখুন
total_vectors = len(collection_data['ids'])
print(f"📊 ডাটাবেজে মোট ভেক্টরের সংখ্যা: {total_vectors} টি\n")
```

```python
# ৩. উদাহরণ হিসেবে প্রথম টুকরোটির আইডি, টেক্সট এবং ভেক্টর এমবেডিং দেখুন
print("=== প্রথম ভেক্টরের বিস্তারিত তথ্য ===")
print(f"🆔 ভেক্টর ID: {collection_data['ids'][0]}")
print(f"📄 মূল টেক্সট (Document): \n{collection_data['documents'][0][:300]}...") # প্রথম ৩০০ অক্ষর দেখাচ্ছে
```

```python
# ৪. ওপেনএআই যে গাণিতিক সংখ্যা (Vector) তৈরি করেছে তা দেখুন
# নোট: ওপেনএআই-এর একেকটি ভেক্টরের দৈর্ঘ্য ১৫৩৬টি সংখ্যার হয়ে থাকে, আমরা প্রথম ১০টি সংখ্যা প্রিন্ট করে দেখছি
print(f"\n🔢 ভেক্টর এমবেডিং (প্রথম ১০টি গাণিতিক সংখ্যা): \n{collection_data['embeddings'][0][:10]}")
print(f"📐 পুরো ভেক্টরের সাইজ/ডাইমেনশন: {len(collection_data['embeddings'][0])}")
```
